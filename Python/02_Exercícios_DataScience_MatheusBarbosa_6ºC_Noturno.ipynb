```python
'''
19. Crie dois arrays de numeros inteiros aleatorios de tamanho 20 e calcule o
angulo entre eles (produto escalar normalizado).
'''
import numpy as np

array1 = np.random.randint(1,10,20)
array2 = np.random.randint(1,10,20)

print("Array 1:")
print(array1)

print("Array 2:")
print(array2)

produto_escalar = np.dot(array1,array2)

norma_array1 = np.linalg.norm(array1)
norma_array2 = np.linalg.norm(array2)

angulo = np.arccos(produto_escalar / (norma_array1 * norma_array2))

print("\nProduto Escalar:", produto_escalar)
print("Angulo entre os arrays em radianos:", angulo)
```

# 2 - Exercıcios de SciPy

```python
'''
1. Calcule a integral da funcao f(x) = x**2 sin(x) no intervalo de 0 a π.
'''
import numpy as np
from scipy.integrate import quad

def f(x):
    return x**2 * np.sin(x)

resultado = quad(f, 0, np.pi)

print(resultado[0])
```

```python
'''
2. Resolva a equacao diferencial dy/dx = x − y, com y(0) = 1, no intervalo de 0
a 10.
'''
import numpy as np
from scipy.integrate import solve_ivp

def f(x, y):
    return x - y

resultado = solve_ivp(f, [0, 10], [1])

print(resultado.y[0])
```

```python
'''
3. Encontre as raızes da funcao f(x) = x**4 − 3x**3 + 2 utilizando SciPy.
'''
import numpy as np
from scipy.optimize import root

def f(x):
  return x**4 - 3*x**3 + 2

resultado = root(f,1)

print("Raiz:", resultado.x[0])
```

```python
'''
4. Use a funcao de otimizacao de SciPy para maximizar f(x) = −x**2+ 4x+ 1
'''
from scipy.optimize import minimize

def f(x):
  return -(x**2) + 4*x + 1

resultado = minimize(lambda x: -f(x), 0)

print("Valor de x:", resultado.x[0])
print("Valor maximo:", f(resultado.x[0]))
```

```python
'''
5. Calcule a transformada de Fourier de um sinal composto de sin(2πt) e
cos(4πt).
'''
import numpy as np
from scipy.fft import fft, fftfreq

t = np.linspace(0,1,1000)

sinal = np.sin(2 * np.pi * t) + np.cos(4 * np.pi * t)

transf = fft(sinal)

frenq = fftfreq (len(t), t[1] - t[0])

for i in range(len(frenq)):
  if abs(transf[i]) > 100:
    print("frenquencia:", frenq[i])
```

```python
'''
6. Encontre a matriz inversa de uma matriz 5x5 de numeros aleatorios.
'''
import numpy as np
from scipy.linalg import inv

matriz = np.random.randint(1,10,(5,5))

inversa = inv(matriz)

print("Matriz Original:")
print(matriz)

print("\nMatriz Inversa:")
print(inversa)
```

```python
'''
7. Use a funcao de interpolacao cubica de SciPy para interpolar os dados
(0, 0),(1, 2),(2, 3),(3, 5),(4, 4).
'''
import numpy as np
from scipy.interpolate import CubicSpline

x = np.array([0, 1, 2, 3, 4])
y = np.array([0, 2, 3, 5, 4])

funcao = CubicSpline(x, y)

novos_x = np.linspace(0, 4, 10)

novos_y = funcao(novos_x)

print("Valores de x:")
print(novos_x)

print("\nValores interpolados:")
print(novos_y)
```

```python
'''
8. Resolva o sistema de equacoes 3x + 2y = 5 e x + 4y = 6.
'''
import numpy as np
from scipy.linalg import solve

A = np.array([
    [3, 2],
    [1, 4]
])

b = np.array([5, 6])

resultado = solve(A, b)

print("x =", resultado[0])
print("y =", resultado[1])
```

```python
'''
9. Ajuste uma curva quadratica aos pontos (1, 2),(2, 4),(3, 5),(4, 7),(5, 8).
'''
import numpy as np
from scipy.optimize import curve_fit

x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 5, 7, 8])

def funcao(x, a, b, c):
    return a*x**2 + b*x + c

parametros, _ = curve_fit(funcao, x, y)

print("a =", parametros[0])
print("b =", parametros[1])
print("c =", parametros[2])
```

```python
'''
10. Calcule a integral dupla da funcao f(x, y) = xy + x**2 no intervalo [0, 2]
para x e [0, 1] para y.
'''
from scipy.integrate import dblquad

def f(y, x):
    return x * y + x**2

resultado = dblquad(f, 0, 2, 0, 1)

print("Resultado:", resultado[0])
```

```python
'''
11. Encontre os autovalores e autovetores de uma matriz 4x4 de numeros
inteiros aleatorios.
'''
import numpy as np
from scipy.linalg import eig

matriz = np.random.randint(1, 10, (4, 4))

autovalores, autovetores = eig(matriz)

print("Matriz:")
print(matriz)

print("\nAutovalores:")
print(autovalores)

print("\nAutovetores:")
print(autovetores)
```

```python
'''
12. Resolva a equacao e**x − x**2 = 0 para encontrar a raiz mais proxima de
x = 1.
'''
import numpy as np
from scipy.optimize import root

def f(x):
    return np.exp(x) - x**2

resultado = root(f, 1)

print("Raiz:", resultado.x[0])
```

```python
'''
14. Utilize a funcao de otimizacao multivariada para minimizar
f(x, y) = (x−2)**2 + (y − 3)**2.
'''
from scipy.optimize import minimize

def f(valores):
    x, y = valores
    return (x - 2)**2 + (y - 3)**2

inicio = [0, 0]

resultado = minimize(f, inicio)

print("x =", resultado.x[0])
print("y =", resultado.x[1])
print("Valor mínimo =", resultado.fun)
```

```python
'''
15. Calcule a transformada de Fourier inversa de um sinal complexo gerado
aleatoriamente.
'''
import numpy as np
from scipy.fft import fft, ifft

sinal = np.random.rand(10) + 1j * np.random.rand(10)

transf = fft(sinal)

resultado = ifft(transf)

print("Sinal original:")
print(sinal)

print("\nTransformada de Fourier:")
print(transf)

print("\nTransformada inversa:")
print(resultado)
```

```python
'''
16. Resolva a equacao diferencial de segunda ordem d**2y/dx**2 +y = 0, com
condicoes iniciais y(0) = 1 e y′(0) = 0.
'''
import numpy as np
from scipy.integrate import solve_ivp

def f(x, y):
    return [y[1], -y[0]]

inicio = [1, 0]

resultado = solve_ivp(f, [0, 10], inicio)

print("Valores de x:")
print(resultado.t)

print("\nValores de y:")
print(resultado.y[0])
```

```python
'''
17. Encontre a raiz cubica de 27 utilizando SciPy.
'''
from scipy.optimize import root_scalar

def f(x):
    return x**3 - 27

resultado = root_scalar(f, bracket=[0, 5])

print("Raiz cúbica de 27:", resultado.root)
```

```python
'''
18. Interpole os valores de uma funcao f(x) = x**2 nos pontos [1, 2, 3, 4, 5]
utilizando spline cubica.
'''
import numpy as np
from scipy.interpolate import CubicSpline

x = np.array([1, 2, 3, 4, 5])

y = x**2

spline = CubicSpline(x, y)

novos_x = np.array([1.5, 2.5, 3.5, 4.5])

novos_y = spline(novos_x)

print("Novos pontos:")
print(novos_x)

print("\nValores interpolados:")
print(novos_y)
```
